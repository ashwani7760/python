```python
import numpy as np
#numpy stands for numerical python
```

```python
# one d array -> vector
arr1d = np.array([10,20,23,20,210,2,12,10])
print(arr1d)
```

```python
# two d array -> matrix
arr2d = np.array([[10,20,30],
                  [40,50,60]])
print(arr2d)
```

```python
#array with defaults values

#zero array or matrix
zero = np.zeros((2,3))
print(zero)
```

```python
#array with one
one = np.ones((2,3))
print(one)
```

```python
#array with given numbers
full = np.full((2,3),7)
print(full)
```

```python
# identity matrix
idm = np.eye(4)
print(idm)
```
